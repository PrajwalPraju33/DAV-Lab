```python
grade_info = pd.DataFrame({
    'grade':['A','B','C','D','F'],
    'grade_descriptiopn':['Excellent','Good','Avegare','Below average','fail']
})

print("\nGrade information DataFrame:")
print(grade_info)
```

```python
merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)

print("\nMerged DataFrame:")
print(merged.head())
```

```python
grade_details=pd.DataFrame({
    'description':['Excellent','Good','Avegare','Below average','fail'],
    'performance_level':['Very high','High','Medium','Low','Very low']},
    index=['A','B','C','D','F'])
print(grade_details.head())
```

```python
grade_counts = df['grade'].value_counts().to_frame('student_count')
print(grade_counts.head())
joined=grade_counts.join(grade_details)
print(joined)
```

```python
df1=df.iloc[:100]
df2=df.iloc[100:200]

combained = pd.concat([df1, df2])
print(combained.head())
print(combained.shape)
```

```python
average_score = df.groupby('grade')['total_score'].mean()
print("\nAvegare total score by grade:")
print(average_score)
```

```python
pivot = pd.pivot_table(
    df,
    values='total_score',
    index='grade',
    aggfunc='mean'
    
)

print('\nPivot Table - Average score by Grade:')
print(pivot)
```
