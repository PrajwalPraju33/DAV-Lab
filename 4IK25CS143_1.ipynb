```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40, 50])

print('Array: ', arr)
print('Mean: ', arr.mean())
print('Reshaped 20:\n', arr.reshape(5, 1))
print('Sliced (index 1 to 3):', arr[1:4])

data = {'Name': ['Asha', 'Ravi', 'Meera', 'karan'],
    'Marks': [89, 72, 91, 65],
    'Branch': ['CSE', 'ECE', 'CSE', 'ISE']}
                

df = pd.DataFrame(data)

print(df)

print('\nFirst 2 rows:\n', df.iloc[0:2])
print('\nMarks column:\n', df['Marks'])
print('\nRows of marks > 70:\n', df[df['Marks'] > 70])
```
